# Processing per capita

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\student\Desktop\Manufacturing_Trends\data\bls.gov\employment\raw\01.csv", delimiter='\t')
df.head()
```

```python
df.value = df.value * 1000
```

```python
df.head()
```

```python
df_pop = pd.read_csv(r"C:\Users\student\Desktop\Manufacturing_Trends\data\census.gov\USpop2020-2025.csv")
df_pop.head()
```

```python
state_codes = pd.read_csv(r"C:\Users\student\Desktop\Manufacturing_Trends\src\Data Gathering\bls.gov\employment\sm.state.txt", delimiter='\t')
state_codes.head()
```

```python
state_codes_dict = {}
for code, name in zip(state_codes.state_code, state_codes.state_name):
    state_codes_dict[code] = name
```

```python
df_pop['STATE_NAME'] = [state_codes_dict[int(state)] for state in df_pop.STATE]
#for state in df_pop.STATE:
#    print(state, state_codes_dict[state])
```

```python
state_set = set(df.state)
```

```python
rows = []

for state in state_set:
    subset = df[df.state == state]
    year_2020 = subset[subset.year == 2020]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2020
    year_2020['per_capita'] = year_2020.value.astype('int') / population
    rows.append(year_2020)

    year_2021 = subset[subset.year == 2021]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2021
    year_2021['per_capita'] = year_2021.value.astype('int') / population
    rows.append(year_2021)

    year_2022 = subset[subset.year == 2022]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2022
    year_2022['per_capita'] = year_2022.value.astype('int') / population
    rows.append(year_2022)

    year_2023 = subset[subset.year == 2023]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2023
    year_2023['per_capita'] = year_2023.value.astype('int') / population
    rows.append(year_2023)

    year_2024 = subset[subset.year == 2024]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2024
    year_2024['per_capita'] = year_2024.value.astype('int') / population
    rows.append(year_2024)

    year_2025 = subset[subset.year == 2025]
    population = df_pop[df_pop.STATE_NAME == state].iloc[0].POPESTIMATE2025
    year_2025['per_capita'] = year_2025.value.astype('int') / population
    rows.append(year_2025)

```

```python
adjusted_per_capita = pd.concat(rows)
adjusted_per_capita.head()
```

```python
adjusted_per_capita.to_csv(r"C:\Users\student\Desktop\Manufacturing_Trends\data\bls.gov\employment\processed\01_per_capita_2020-2025.csv")
```
