# Adjusting Robotics Investment Data to be Per Worker

```python
import pandas as pd
```

### Load and Adj employment data

```python
employment_data = pd.read_csv(r"..\data\bls.gov\employment\raw\01.csv", delimiter='\t')
employment_data.head()
```

```python
employment_data = employment_data[employment_data.year == 2022]
employment_data.head()
```

```python
employment_data_by_state = employment_data.groupby('state').value.sum()
employment_data_by_state.head()
```

```python
employment_data_dict = employment_data_by_state.to_dict()

USA_total = sum(employment_data_dict.values())
employment_data_dict['United States'] = USA_total
```

### Load Robotics Investment Data

```python
investment_data = pd.read_csv(r"..\data\census.gov\robotics_ce_by_state_2022.csv", delimiter='\t')
investment_data.columns = ["State", "2022_Robotics_CE"]
investment_data['2022_Robotics_CE'] = investment_data["2022_Robotics_CE"].astype('float')
investment_data.head()
```

# Calculating Adjusted CE

```python
investment_data['CE_adjusted'] = investment_data['2022_Robotics_CE'] / investment_data['State'].map(employment_data_dict)
investment_data.head()
```

```python
investment_data.to_csv(r"..\data\census.gov\processed\robotics_ce_by_state_2022_adjusted.csv", index=False)
```
